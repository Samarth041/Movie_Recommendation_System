```python
import pandas as pd
import numpy as np
import joblib
```

```python
movies = pd.read_parquet(
    "../data/processed/movies_features.parquet"
)

similarity_matrix = joblib.load(
    "../data/processed/content_similarity.pkl"
)

movie_indices = joblib.load(
    "../data/processed/movie_indices.pkl"
)

svd_model = joblib.load(
    "../data/processed/svd_model.pkl"
)
```

suppose user has given a movie name then it will find top 50 movie which is similar to it  and then it will send it getrank_candidate which will further personalise that will he like it or not

```python
#content candidates

def get_content_candidates(title,top_n=50):
    idx=movie_indices[title]

    similarity_scores=list(enumerate(similarity_matrix[idx]))

    similarity_scores=sorted(
        similarity_scores,
        key=lambda x:x[1],
        reverse=True
    )[1:top_n+1]

    movie_ids=[
        i[0]
        for i in similarity_scores
    ]

    return movies.iloc[movie_ids]
```
